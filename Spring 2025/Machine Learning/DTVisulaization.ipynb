```python
# Import necessary libraries
import numpy as np  # For numerical operations
import pandas as pd  # For data manipulation
from sklearn.datasets import load_diabetes  # For loading the diabetes dataset
import seaborn as sns  # For data visualization
import matplotlib.pyplot as plt  # For plotting charts

# Step 1: Load the Diabetes dataset from scikit-learn
data = load_diabetes()

# Step 2: Get the feature data (X) and target variable (y)
X = data.data  # Features (input variables)
y = data.target  # Target variable (diabetes progression)

# Step 3: Convert the features to a pandas DataFrame for easier exploration
df = pd.DataFrame(X, columns=data.feature_names)  # Create DataFrame with column names from the dataset

# Step 4: Add the target variable to the DataFrame to complete the dataset
df['Target'] = y  # Add target (diabetes progression) as a column in the DataFrame

# Step 5: Display the first 5 rows of the dataset
print("First 5 rows of the dataset:")
print(df.head())  # Shows the first 5 rows of the DataFrame

# Step 6: Display summary statistics of the dataset
print("\nSummary statistics of the dataset:")
print(df.describe())  # Displays the summary statistics like mean, std, min, max, and quartiles

# Step 7: Display information about the dataset (data types, non-null values)
print("\nDataset Information:")
print(df.info())  # Shows data types and non-null count for each column

# Step 8: Check for any missing values in the dataset
print("\nMissing Values:")
print(df.isnull().sum())  # Checks and sums the missing values in each column

# Step 9: Visualize the distribution of the target variable (diabetes progression)
plt.figure(figsize=(8,6))  # Set the figure size
sns.histplot(df['Target'], kde=True, color='skyblue')  # Plot a histogram with a KDE (kernel density estimate)
plt.title('Distribution of Diabetes Target Variable')  # Title of the plot
plt.xlabel('Diabetes Progression')  # X-axis label
plt.ylabel('Frequency')  # Y-axis label
plt.show()  # Display the plot

# Step 10: Visualize the correlation between features using a heatmap
plt.figure(figsize=(10, 8))  # Set the figure size for the heatmap
corr_matrix = df.corr()  # Calculate the correlation matrix for all features
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)  # Heatmap with correlation values
plt.title('Correlation Heatmap of Features')  # Title of the heatmap
plt.show()  # Display the heatmap

# Step 11: Visualize pairwise relationships between features using a pairplot
sns.pairplot(df, hue='Target', palette='coolwarm')  # Pairplot with different colors based on target variable
plt.show()  # Display the pairplot

```
