# Modifying an Image
Taking two images, we will be able to first show them.
After this, our second assignment is to change in image from RGB to grayscale.
Finally, we must Change any one-fourth of the image into black.

```python
from PIL import Image
```

### Showing the images

```python
flower = Image.open('JL-gardinia.jpg')
#flower.show()

bull = Image.open('JL-bull.jpg')
#bull.show()
```

### Changing the Images from RGB to GrayScale

```python
#using mode L
flower.convert(mode='L').save('greyFlower.jpg')
gFlow = Image.open('greyFlower.jpg')
bull.convert(mode='L').save('greyBull.jpg')
gBull = Image.open('greyBull.jpg')
```

### Changing 1/4 of the Image to Black

```python
def fourthBlack(image):
    #get the middle of the picture
    width, height = image.size
    midW = int(0.5 * width)
    midH = int(0.5 * height)
    
    #now while loop to get one-fourth black
    black_pixel = (0, 0, 0)
    for x in range(midW):
        for y in range(midH):
            #image[x, y] = (0, 0, 0)
            image.putpixel((x, y), 0)
```

```python
#blackBull = fourthBlack(gBull).save('blackBull.jpg')
blackBull = gBull.copy()
fourthBlack(blackBull)
blackBull.save('blackBull.jpg')

blackFlower = gFlow.copy()
fourthBlack(blackFlower)
blackFlower.save('blackFlower.jpg')
```
