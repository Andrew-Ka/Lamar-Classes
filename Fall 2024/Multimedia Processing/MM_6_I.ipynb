### Image Flip
Easiest way of doing this is just swapping the columns. This can be done since this is a numpy array.    
Credits: <a href="https://stackoverflow.com/questions/58814369/flipping-an-image-vertically-in-python-without-using-flip-method">link</a>

```python
#flip
image = np.array([list(reversed(row)) for row in image])
cv2_imshow(image)
```

### Convolutional


```python
#making the kernel
kernelEdge = np.array([[0, 1, 0],
                       [1, -4, 1],
                       [0, 1, 0]])
#applying the kernel filter
filtImg = cv2.filter2D(image, -1, kernelEdge)
#show the image
cv2_imshow(filtImg)
```
